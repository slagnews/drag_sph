```python
EMPTY = -1

# Example values (you can replace with actual values)
n_atoms = 1000
Lx, Ly = 10.0, 10.0          # Size of the domain
rc = 1.0                     # Cutoff / cell size
nx = int(Lx / rc) # amount of cells in x direction
ny = int(Ly / rc) # amount of cells in y direction
ncxy = nx * ny # Total number of cells

# Arrays for positions
positions = np.random.rand(n_atoms, 2) * [Lx, Ly]

# Initialize data structures
lscl = np.full(n_atoms, EMPTY, dtype=int)
head = np.full(ncxy, EMPTY, dtype=int)
```

```python
def vec_to_scalar_index(ix, iy):
    return ix * ny + iy

def get_cell_index(pos):
    cx = int(pos[0] / rc)
    cy = int(pos[1] / rc)
    return cx, cy
```

```python
def construct_linked_list():
    global head, lscl
    head[:] = EMPTY
    for i in range(n_atoms):
        cx, cy = get_cell_index(positions[i])
        c = vec_to_scalar_index(cx, cy)
        lscl[i] = head[c]
        head[c] = i
```

```python
construct_linked_list()

print(positions[0])

cell_index = get_cell_index(positions[0])
print(cell_index)

scalar_index = vec_to_scalar_index(*cell_index)
print(scalar_index)

first_particle = head[scalar_index]
print(first_particle)
    
next_particle = lscl[first_particle]
print(positions[next_particle])
```
